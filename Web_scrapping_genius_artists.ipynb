```python
import requests   
from bs4 import BeautifulSoup  

artist_slug = "Ed-sheeran" 
artist_name = "Ed-sheeran" 
artist_url = f"https://genius.com/artists/{artist_slug}"  

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
   
}

response = requests.get(artist_url, headers=headers)  # Отправляем запрос с заголовками
response.raise_for_status()  # Проверяем статус ответа; если ошибка - возбуждаем исключение


with open("genius_artist_page.html", "w", encoding="utf-8") as f: 
    f.write(response.text) 
print("Сохранён исходный HTML в файл: genius_artist_page.html\n") 


soup = BeautifulSoup(response.text, 'html.parser')  


# Выводим информацию об артисте
print(f"Артист: {artist_name}")  
print(f"Страница: {artist_url}\n")  

songs = []  

song_links = soup.select('a.mini_card')[:3]  

for link in song_links:
    title = link.get_text(strip=True)  
    href = link.get('href')  
    full_link = "https://genius.com" + href if href and href.startswith('/') else href  
    songs.append({"title": title, "link": full_link})  

if songs:
    print("Первые 3 популярные песни:")  
    for i, song in enumerate(songs, 1):
        print(f"   {i}. {song['title']}")  
        print(f"      Ссылка: {song['link']}")  
else:
    print("Песни не найдены.")
```

```python
import requests   
from bs4 import BeautifulSoup  

# Настройки артиста для формирования URL и вывода имени
artist_slug = "Ed-sheeran"  # часть URL артиста на сайте genius.com, используется для построения адреса
artist_name = "Ed-sheeran" 
artist_url = f"https://genius.com/artists/{artist_slug}"  

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
   
}

# Выполняем HTTP GET-запрос на страницу артиста
response = requests.get(artist_url, headers=headers)  # Отправляем запрос с заголовками
response.raise_for_status()  # Проверяем статус ответа; если ошибка - возбуждаем исключение

# Сохраняем текст HTML-страницы в файл для проверки и отладки
with open("genius_artist_page.html", "w", encoding="utf-8") as f:  # Открываем файл для записи с UTF-8
    f.write(response.text)  # Записываем полученный HTML в файл
print("Сохранён исходный HTML в файл: genius_artist_page.html\n")  # Выводим сообщение о сохранении файла

# Создаем объект BeautifulSoup для парсинга HTML-кода страницы
soup = BeautifulSoup(response.text, 'html.parser')  
# 'html.parser' — встроенный парсер Python, подходящий для большинства HTML документов

# Выводим информацию об артисте
print(f"Артист: {artist_name}")  # Выводим имя артиста (без лишних пробелов)
print(f"Страница: {artist_url}\n")  # Выводим URL страницы артиста с переносом строки для читабельности

# Создаем пустой список для хранения данных о песнях
songs = []  

# Находим все теги <a> с классом 'mini_card' — это ссылки на песни
song_links = soup.select('a.mini_card')[:3]  # Берем первые 3 ссылки, чтобы ограничить вывод

# Цикл по найденным песням для сбора информации
for link in song_links:
    title = link.get_text(strip=True)  
    # Извлекаем текст из тега, убирая пробелы и переносы по краям
    href = link.get('href')  # Получаем значение атрибута href — ссылку на песню
    # Проверяем, начинается ли ссылка с '/', если да — формируем полный URL
    full_link = "https://genius.com" + href if href and href.startswith('/') else href  
    # Добавляем словарь с названием песни и её ссылкой в список
    songs.append({"title": title, "link": full_link})  

# Проверяем, есть ли найденные песни в списке
if songs:
    print("Первые 3 популярные песни:")  # Заголовок для списка песен
    # Проходим по всем песням с нумерацией, начиная с 1
    for i, song in enumerate(songs, 1):
        print(f"   {i}. {song['title']}")  # Выводим номер и название песни с отступом
        print(f"      Ссылка: {song['link']}")  # Выводим URL песни с большим отступом
else:
    # Если список песен пуст, выводим сообщение об отсутствии результата
    print("Песни не найдены.")
```
